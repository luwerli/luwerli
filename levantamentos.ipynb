```python
## Script that convert local coordinates to utm coordinates
## considering that we dont know the epsg of these coordinates.

import pyproj
import math

def read_coordinates(file_path):
    with open(file_path, 'r') as f:
        lines = f.readlines()
    coords = []
    id = []
    lat = []
    lon = []
    for line in lines[1:]:  # Skip header
            parts = line.strip().split(',')
            if len(parts) >= 3:
                try:
                    id.append(parts[0])
                    lat.append(float(parts[1]))
                    lon.append(float(parts[2]))
                    coords.append((id, lat, lon))
                except ValueError:
                    print(f"Invalid line skipped: {line.strip()}")
    return coords, lat, lon

def convert_to_utm(lat, lon):
    utm_zone = int((lon + 180) / 6) + 1
    is_northern = lat >= 0
    proj_utm = pyproj.Proj(proj='utm', zone=utm_zone, ellps='GRS80', south=not is_northern)
    easting, northing = proj_utm(lon, lat)
    return easting, northing, utm_zone

def save_utm_coordinates(coords, output_file):
    with open(output_file, 'w') as f:
        f.write("ID,Easting,Northing,Zone\n")  # Optional header
        for id, lat, lon in coords:
            easting, northing, zone = convert_to_utm(lat, lon)
            f.write(f"{id},{easting:.3f},{northing:.3f},Zone {zone}\n")

def convert_to_geodetic(lat, lon):
    lat_geo = []
    lon_geo = []
    lat_geo.append.math.atan
    lon_geo = math.atan(lat/lon)
    return lat_geo, lon_geo

def save_geodetic_coordinates(coords, output_file):
    with open(output_file, 'w') as f:
        f.write("ID,Latitude,Longitude,\n")  # Optional header
        for id, lat, lon in coords:
            latitude, longitude, altitude = convert_to_utm(lat, lon)
            f.write(f"{id},{latitude:.3f},{longitude:.3f},Zone {altitude}\n")


def main():
    input_file = r'C:\Users\luiza\OneDrive - UFRGS\SEMESTRES\2025.1\NILULU.txt'
    output_UTM = r'C:\Users\luiza\OneDrive - UFRGS\SEMESTRES\2025.1\NILULU_UTM.txt'
    output_geodetic = r'C:\Users\luiza\OneDrive - UFRGS\SEMESTRES\2025.1\NILULU_geodetic.txt'
    coords, lat, lon = read_coordinates(input_file)
    #coords = read_coordinates(input_file)
    save_utm_coordinates(coords, output_UTM)

if __name__ == "__main__":
    main()
```
